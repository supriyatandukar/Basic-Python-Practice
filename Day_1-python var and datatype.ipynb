# Rules for Python variables:

### ->Cannot be any of the Python keywords.
### ->Cannot start with a number
### ->Start with a letter or the underscore character
### ->Can only contain alpha-numeric characters and underscores (A-z, 0-9, and _ )
### ->Case-sensitive (age, Age and AGE are three different variables)

## 6. Challenge: The "Data Type" Mini-Project

```python
import datetime

name = input("Write your name.")
birth = int(input("Write your birth year."))
height = float(input("Write your height in metre."))

current_year = datetime.date.today().year
age = current_year - birth
is_tall = height > 1.8

print(f"User {name} is {age} years old. You are {height} meters tall. It is {is_tall} that you are over 1.8 meters.")
```
