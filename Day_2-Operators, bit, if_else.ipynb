```python
a = 15
b = 4

print(f"a = {a}, b = {b}")
print(f"Addition (a + b): {a + b}")           # Addition
print(f"Subtraction (a - b): {a - b}")        # Subtraction
print(f"Multiplication (a * b): {a * b}")     # Multiplication

print(f"Division (a / b): {a / b}")           # Division (returns float)
print(f"Floor Division (a // b): {a // b}")   # Floor Division (returns integer)
print(f"Modulus (a % b): {a % b}")            # Modulus (remainder)
print(f"Exponentiation (a ** b): {a ** b}")   # Exponentiation (a to the power b)
```

```python
print("\n--- More Examples ---")

# Floor division vs regular division
print(f"10 / 3 = {10 / 3}")
print(f"10 // 3 = {10 // 3}")
```

```python
# Modulus operator for remainder
print(f"17 % 5 = {17 % 5}")  

# Exponentiation
print(f"2 ** 3 = {2 ** 3}")  
print(f"5 ** 2 = {5 ** 2}")

# Negative numbers
print(f"-10 // 3 = {-10 // 3}")
print(f"-10 % 3 = {-10 % 3}")
```

```python
# Example 4: Conditions with logical operators (and, or, not)
has_ticket = True
has_id = False
print(f"Has Ticket: {has_ticket}, Has ID: {has_id}")

if has_ticket and has_id:
    print("Entry allowed.")
elif has_ticket or has_id:
    print("Partial requirements met. Check with staff.")
    if not has_id:
        print("Warning: ID is missing!")
else:
    print("Entry denied.")
```
