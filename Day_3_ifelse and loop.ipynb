### 2.1 `for` Loops

`for` loops are used to iterate over a *sequence* (like a list, tuple, dictionary, set, or string) or other *iterable* object.

**Syntax:**

```python

    # Code block to execute for each item in the iterable
    # 'variable' takes the value of the current item in each iteration
    print(variable)

````python
number = [1,2,3,4,10,6, 4, 66, 77]
string = """Global web icon

We would like to show you a description …

"""

for value in string:
    print(value)
````

`while` loops execute a block of code *as long as* a specified condition remains `True`.

**Syntax:**

```python
while a>10:
    # Code block to execute as long as condition is True
    print("Condition is still True")
    # IMPORTANT: Usually, something inside the loop needs to
    # change the condition, otherwise it might loop forever!

````python
count = 0
print("Counting up to 5 using while loop:")
while count < 5:
    print(f"Count is: {count}")
    count += 1

print("Loop finished.")
print("-" * 20)
````

## 3. Loop Control Statements (`break`, `continue`, `pass`)

These statements allow you to modify the standard behavior of loops.

*   **`break`**: — Stop the loop right now!
*   **`continue`**: Skip this one and move on.
*   **`pass`**: Does nothing. It acts as a placeholder where syntax requires a statement, but you don't want any code to execute. This is often used in empty function definitions, classes, or conditional blocks during development.

````python
# Example 1: break
print("Finding the first multiple of 7 between 1 and 20:")
lists = []
for number in range(1, 21):
   
    if number % 7 == 0:
        print(f"Found it! {number} is a multiple of 7.")
        break
    else:
        lists.append(number)

print("Loop finished or broken.")
print("Lists:", lists)
````

````python
# Example 1: break
print("Finding my bias.......")

nexz = ["Yu", "Haru", "Soken", "Seita", "Hyui", "Yuki", "Moya", "Yu", "Haru", "Moya", "Soken", "Seita", "Hyui", "Yuki"]
lists = []

for leader in nexz :   
    if leader == "Moya":
        print(f"Found u my bias leader! {leader}")
        break #stops searching after finding one
        
    else:
        lists.append(leader)

print("List before bias:", lists)
````

### Activity 1: Number Guessing Game

**Goal:** The computer picks a secret number, and the user has to guess it. The program gives hints like "too high" or "too low".

**Concepts We'll Use:**
*   `import random` (to get a random number)
*   `input()` (to get the user's guess)
*   `int()` (to convert the user's input string into a number)
*   `while` loop (to keep guessing until correct)
*   `if`/`elif`/`else` (to check the guess and give hints)

````python
import random
num = random.randint(0, 100) #this means any random num from 0 to 100
guess = 0
count = 0

print("Guess the number from 1 to 100.:")
while guess != num:
    guess = int(input("Guess any number: "))
#yeta ni count ta rakhna milthyo but correct guess lai ni count gardinxa so ki ta count = 1 garnu paryo ki ta if, elif ma aaile rakhya jasari rakhne    
    if guess < num:  #Wrote <= here first so it means less and equal meaning equal always show true so it didn;t work
        print("Not correct, try guessing higher.")
        count += 1
    elif guess > num:
        print("Not correct, try guessing lower.")
        count += 1
    else:
        print(f"Correct!You attempted {count} times to make a correct guess.")
````
