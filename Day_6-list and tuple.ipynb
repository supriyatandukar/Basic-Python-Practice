### List Comprehensions

```python
#Paila list banaune ani kaam k ho vanera describe garne ani feri condn deko xa
#operation, loop, filter ek line mai
numbers = [1,2,4,5,6,8,9]

squares = [x**2  for x in numbers if x%2 ==0]  #yo vaneko numbers ko square ma 2 le divide garda reminder 0 aaune maytra aayeko

print(squares)
```

```python
#mathi ko yesari ni garna paii
y = []
for x in numbers:
    if x%2==0:
        z=x**2
        y.append(z)

print(y)
```

```python
#List comprehension
s = "Learning python programming"
vowels = [char for char in s if char in "aeiou"] #for (char) ma name j rakhe ni hunxa
print(vowels, end="")
```

```python
#repeat huna nadina set banauna ni payo
```

```python
s = "Learning python programming"
vowels = [char for char in s if char not in "aeiou"] #membership operator
print(vowels, end="")
```
