# Control Flow — Conditionals

3. Build a number classifier: positive/negative/zero,
odd/even
4. Write a ticket pricing system: child (<10), student
(10-18), adult (18-60), senior (60+)
5. Create a simple ATM simulator: check balance, warn
if low

---

```python
number = int(input("Enter a number: "))

# Positive, Negative, or Zero
if number > 0:
    print("The number is Positive.")
elif number < 0:
    print("The number is Negative.")
else:
    print("The number is Zero.")

# Odd or Even
if number % 2 == 0:
    print("The number is Even.")
else:
    print("The number is Odd.")
```

```python
age = int(input("Enter your age: "))

if age < 10:
    price = 100
    category = "Child"
elif age < 18:
    price = 150
    category = "Student"
elif age < 60:
    price = 250
    category = "Adult"
else:
    price = 120
    category = "Senior"

print("\n=== Ticket Information ===")
print(f"Category: {category}")
print(f"Ticket Price: Rs. {price}")
```

```python
balance = 500

print("=== ATM Simulator ===")
print(f"Current Balance: Rs. {balance}")

if balance < 100:
    print("Warning: Your balance is running low.")
else:
    print("Your balance is sufficient.")
```

```python
balance = float(input("Enter your account balance: Rs. "))

print("\n=== ATM Simulator ===")
print(f"Current Balance: Rs. {balance:.2f}")

withdraw = float(input("Enter amount to withdraw: Rs. "))

if withdraw <= balance:
    balance -= withdraw
    print(f"\nWithdrawal Successful!")
    print(f"Remaining Balance: Rs. {balance:.2f}")

    if balance < 100:
        print("Warning: Low Balance!")
else:
    print("\nInsufficient Funds!")
```

# Control Flow — Loops
6. Print the multiplication of any number
7. Print first 30 Fibonacci numbers using a loop
8. Write a loop that finds all prime numbers between 1
and 100
9. Build a simple number-guessing game using while
loop

---

```python
num = int(input("Enter a number: "))

for i in range(1, 11):
    print(f"{num} x {i} = {num*i}")
```

```python
a, b = 0, 1

for _ in range(30):
    print(a, end=" ")
    a, b = b, a + b
```

```python
for num in range(2, 101):
    is_prime = True

    for i in range(2, int(num**0.5) + 1):
        if num % i == 0:
            is_prime = False
            break

    if is_prime:
        print(num)
```

```python
secret = 7

while True:
    guess = int(input("Guess the number: "))

    if guess == secret:
        print("Correct!")
        break
    elif guess < secret:
        print("Too low!")
    else:
        print("Too high!")
```

## Data Structures — Lists & Tuples
1. Flatten a nested list of Nepal districts by province
using list comprehension
2. Write a function that returns top-N items from a list
sorted by value
3. Use tuple unpacking to swap values and unpack
coordinates from a list of GPS points

---

```python
districts = [
    ["Kathmandu", "Bhaktapur"],
    ["Pokhara", "Gorkha"],
    ["Biratnagar", "Jhapa"]
]

flat_list = [district for province in districts for district in province]

print(flat_list)
```

```python
def top_n(numbers, n):
    return sorted(numbers, reverse=True)[:n]

print(top_n([5, 10, 2, 8, 15], 3))
```

```python
x, y = 10, 20
x, y = y, x

print(x, y)

gps_points = [(28.20, 83.98), (27.70, 85.32)]

for lat, lon in gps_points:
    print(f"Latitude: {lat}, Longitude: {lon}")
```

# Dictionaries, Sets & File Handling
1. Read a CSV of NEPSE stock prices; find max, min,
average closing price
2. Write a word frequency counter that reads a .txt file
and outputs top 10 words
3. Parse NRB exchange rate JSON (mock data) and
format as a readable table

---

```python
import pandas as pd

df = pd.read_csv("stock_prices.csv")

print("Max:", df["Close"].max())
print("Min:", df["Close"].min())
print("Average:", df["Close"].mean())
```

```python
from collections import Counter

with open("sample.txt", "r") as file:
    words = file.read().lower().split()

counter = Counter(words)

print(counter.most_common(10))
```

```python
import json

with open("rates.json") as file:
    data = json.load(file)

for currency, rate in data.items():
    print(f"{currency:<10} {rate}")
```
